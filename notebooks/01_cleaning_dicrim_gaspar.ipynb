```python
df_dicrim['cod_commune'] = df_dicrim['cod_commune'].astype(str).str.split('.').str[0].str.zfill(5)
```

```python
if 'dat_publi_dicrim' in df_dicrim.columns:
    df_dicrim['dat_publi_dicrim'] = pd.to_datetime(df_dicrim['dat_publi_dicrim'], errors='coerce')
```

```python
# Remplacement des accents et passage en minuscules en Python (équivalent de ton code SQL)
df_dicrim['lib_commune'] = (
    df_dicrim['lib_commune']
    .astype(str)
    .str.lower()
    .str.replace('é', 'e', regex=False)
    .str.replace('è', 'e', regex=False)
    .str.replace('ê', 'e', regex=False)
    .str.replace('à', 'a', regex=False)
    .str.replace('â', 'a', regex=False)
    .str.replace('ï', 'i', regex=False)
    .str.replace('î', 'i', regex=False)
    .str.replace('ô', 'o', regex=False)
)

print("Nettoyage des accents et minuscules effectué avec succès !")
display(df_dicrim[['lib_commune']].head())
```
